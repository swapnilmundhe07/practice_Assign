# Word2Vec

+ we will use `gensim` used for topic modeling
+ It has already pretrained model which can used to genrating embbeddings,performing similarities etc
+ we can fine tune it 
+ first install gensim : `pip install gensim` 
+ Word2Vec is not contextual embedding it just a word embedding -> `fast,glove,gensim` -> it give same embedding for all words in all context.


```python
import numpy as np
from gensim.models import Word2Vec
```

### get data

```python
# get data
sentences = [
    ["i",'love','python'],
    ['python','is','easy'],
    ['i','love','machine','learning'],
    ['deep','learning','is','fun'],
    ['i','deposited','money','in','bank'],
    ['i','sat','down','near','river','bank']
]
```

### train word2vec model on our data

```python
# create the model based on data
model = Word2Vec(
    # our dataset used to finetune of our model
    sentences=sentences,

    # size of each word embedding
    vector_size=5,

    # no of workers
    workers = 1,

    # context windows size
    window=3,

    # use skipgram
    sg =1,

    # how many times a word must appear in your training data before Word2Vec includes it in the vocabulary.
    min_count=1 # mandatory parameter
)
```

### vocab


```python
# get vocab of words
print(model.wv.index_to_key)

#
```

### get word embedding

```python
## get embedding of word python
model.wv['python']
```

```python
# for both bank word embedding is same not different embedding (contextual embedding)
# gpt uses positional encoding for contextual meaning and order of words
# now ROPE(Rotetary posiitonal encoding) uses gpt models 
# most of model uses ROPE now a days
model.wv['bank']
```

### find similar words

```python
# find similar word to python
similar_word = model.wv.most_similar('python',topn=4)
print(similar_word)
```
